### Feature Importance with Statistical Tests

#### Checking Normality of the dataset

```python
from scipy.stats import shapiro

# Split the data into normal and anomaly groups
normal_data = financial_data_df[financial_data_df['Y'] == 0].drop(columns='Y')
anomaly_data = financial_data_df[financial_data_df['Y'] == 1].drop(columns='Y')

# Initialize results dictionaries
normality_results = {}
mannwhitney_results = {}

print("Normality Test Results (Shapiro-Wilk)")

# normality tests (Shapiro-Wilk)
for column in normal_data.columns:
    stat_normal, p_normal = shapiro(normal_data[column])
    stat_anomaly, p_anomaly = shapiro(anomaly_data[column])
    normality_results[column] = {
        "Normal Group (Y=0)": p_normal,
        "Anomaly Group (Y=1)": p_anomaly,
        "Normality in Both Groups": (p_normal > 0.05) and (p_anomaly > 0.05)
    }



# Convert results to DataFrames for better visualization
normality_df = pd.DataFrame(normality_results).T
normality_df
```

- For every feature, the p-value is far below 0.05 for both Y=0 (Normal Group) and Y=1 (Anomaly Group).
- This means that none of the features are normally distributed in either group.


Based on the results of the normality test, the data does not follow a normal distribution. This finding justifies the use of the Mann-Whitney U Test, a non-parametric statistical method, as a more appropriate alternative to the T-test for further analysis.

#### Mann-Whitney U Test

```python
from scipy.stats import mannwhitneyu

print("Mann-Whitney U Test Results")

for column in normal_data.columns:
    stat, p_value = mannwhitneyu(normal_data[column], anomaly_data[column], alternative='two-sided')
    mannwhitney_results[column] = {
        "Statistic": stat,
        "P-Value": p_value,
        "Significant (P < 0.05)": p_value < 0.05
    }

mannwhitney_df = pd.DataFrame(mannwhitney_results).T

mannwhitney_df
```

```python
# significant features
significant_true = mannwhitney_df[mannwhitney_df["Significant (P < 0.05)"] == True]

# Rank the significant features based on their p-values in ascending order
ranked_significant_features = significant_true.sort_values(by="P-Value").reset_index()

# Display the ranked significant features
print("Significant Features - Ranked Significant Features (P < 0.05):")
print(ranked_significant_features)
```

```python
significant_features = ranked_significant_features['index'].tolist()

print("Significant Features:")
print(significant_features)
print(len(significant_features))
```

```python
# non-significant features
significant_false = mannwhitney_df[mannwhitney_df["Significant (P < 0.05)"] == False].index.tolist()

print("\nNon-significant Features - with Significant (P < 0.05) = False:")
print(significant_false)
```

### Selecting best features for modelling

#### Combine Feature Lists

Combine the features from both methods, removing duplicates while prioritizing those ranked higher in either method.

```python
combined_features = list(dict.fromkeys(correlations_features + significant_features))
print(combined_features)
```

#### Rank Features by Importance

- Assign weights to features based on their significance in the two methods:
  - Correlation with Target: Higher absolute correlation scores indicate stronger linear relationships.
  - Mann-Whitney U Test: Lower p-values indicate stronger distributional differences.

```python
# Create a dictionary for feature ranking
feature_scores = {}

# Add correlation scores (normalized by absolute value)
for feature in correlations_features:
    feature_scores[feature] = 1 / (1 + correlations_features.index(feature))  # Higher rank, higher score

# Add Mann-Whitney scores (normalized by rank in ascending order of p-value)
for feature in significant_features:
    if feature in feature_scores:
        feature_scores[feature] += 1 / (1 + significant_features.index(feature))
    else:
        feature_scores[feature] = 1 / (1 + significant_features.index(feature))

# Convert to sorted feature list based on scores
ranked_features = sorted(feature_scores.items(), key=lambda x: x[1], reverse=True)
ranked_features = [feature[0] for feature in ranked_features]
print(ranked_features)
```

#### Remove Redundant or Highly Correlated Features

- Use correlation analysis among the selected features to remove redundancy:

  - Compute a correlation matrix for ranked_features.
  - Drop one feature from pairs with high correlation (e.g., |correlation| > 0.9).

```python
# Calculate correlation matrix for selected features
selected_data = financial_data_df[ranked_features]
correlation_matrix = selected_data.corr()

# Create a mask to hide the diagonal
mask = np.triu(np.ones_like(correlation_matrix, dtype=bool))

# Plot the heatmap with the mask
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, mask=mask, cmap="coolwarm", annot=False, fmt=".2f", square=True)
plt.title("Correlation Heatmap for selected features (Diagonal Hidden)")
plt.show()
```

```python
# Identify highly correlated pairs
high_correlation_pairs = []
for i in range(len(correlation_matrix.columns)):
    for j in range(i):
        if abs(correlation_matrix.iloc[i, j]) > 0.9:
            high_correlation_pairs.append((correlation_matrix.columns[i], correlation_matrix.columns[j]))

# Drop redundant features (keep only one feature from each pair)
redundant_features = set(pair[1] for pair in high_correlation_pairs)
final_features = [feature for feature in ranked_features if feature not in redundant_features]
print(final_features)
print(len(final_features))
```

```python
# !pip install --upgrade xgboost scikit-learn # Upgrade xgboost and scikit-learn
# !pip install scikit-learn==1.2.2
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, f1_score
from sklearn.model_selection import GridSearchCV

# Separate features and target for both train and test sets
X_train = train_set.drop(columns=['Y'])
y_train = train_set['Y']
X_test = test_set.drop(columns=['Y'])
y_test = test_set['Y']

# reg_alpha is L1 
# reg_lambda is L2
# Define the parameter grid for GridSearchCV
param_grid = {
    'reg_alpha': [0, 0.1, 0.5, 1, 5, 10],  # Example values for reg_alpha
    'reg_lambda': [0, 0.1, 1, 5, 10, 100]   # Example values for reg_lambda
}

# Create the XGBoost model
xgb_model = XGBClassifier(random_state=42, eval_metric='logloss')

# Define a scoring function for f1-score
def f1_scorer(estimator, X, y):
    y_pred = estimator.predict(X)
    return f1_score(y, y_pred)  

# Perform grid search using the scoring function
grid_search = GridSearchCV(estimator=xgb_model, param_grid=param_grid, scoring=f1_scorer, cv=5)
grid_search.fit(X_train, y_train)

# Print the best parameters
print("Best parameters:", grid_search.best_params_)

# Train the model with the best parameters
xgb_model_best = XGBClassifier(**grid_search.best_params_) 
xgb_model_best.fit(X_train, y_train)

# Make predictions on the test set
y_test_pred = xgb_model_best.predict(X_test)

# Make predictions on both train and test sets
y_train_pred = xgb_model_best.predict(X_train)
# y_test_pred = xgb_model.predict(X_test)

# Evaluate performance on train set
print("Classification Report - Train Set:")
print(classification_report(y_train, y_train_pred))

# Evaluate performance on test set
print("\nClassification Report - Test Set:")
print(classification_report(y_test, y_test_pred))
```

```python
# Separate features and target
X_train = train_set.drop(columns=['Y'])
y_train = train_set['Y']
X_test = test_set.drop(columns=['Y'])
y_test = test_set['Y']

# Normalize the data using StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train the Isolation Forest model
iso_forest = IsolationForest(contamination=0.1, random_state=42)  # Adjust contamination if needed
iso_forest.fit(X_train_scaled)

# Predict anomalies (-1 = anomaly, 1 = normal)
y_train_pred = iso_forest.predict(X_train_scaled)
y_test_pred = iso_forest.predict(X_test_scaled)

# Convert predictions to match Y format (0 = normal, 1 = anomaly)
y_train_pred = [0 if pred == 1 else 1 for pred in y_train_pred]
y_test_pred = [0 if pred == 1 else 1 for pred in y_test_pred]

# Evaluate performance
print("Classification Report - Train Set:")
print(classification_report(y_train, y_train_pred))

print("\nClassification Report - Test Set:")
print(classification_report(y_test, y_test_pred))

# Optional: Analyze anomaly scores
anomaly_scores = iso_forest.decision_function(X_test_scaled)
```
